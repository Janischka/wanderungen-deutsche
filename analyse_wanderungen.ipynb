```python
import pandas as pd
```

```python
# Daten aus den 90ern einlesen
df_90er = pd.read_csv("daten/wanderungen_1991-1999.csv")
print(df_90er)
```

```python
# Nur die benötigten Spalten zu Deutschen in neuem df speichern
df_90er = df_90er[["Unnamed: 0", "Deutsche", "Deutsche.1", "Deutsche.2"]]
print(df_90er)
```

```python
# Erste Datenzeile zu Spaltennamen machen und dann Doppelung entfernen
df_90er.columns = df_90er.iloc[0]
df_90er = df_90er.iloc[1:].reset_index(drop=True)
df_90er.columns.name = None
print(df_90er)
```

```python
# Datatypes der Spalten checken 
print(df_90er.dtypes)
```

```python
# Leerzeilen in Angaben entfernen und in numeric umwandeln
for spalte in ["Zuzüge", "Fortzüge", "Saldo"]:
    df_90er[spalte] = pd.to_numeric(
        df_90er[spalte].astype(str).str.replace(r"\s+", "", regex=True)
    )

# Auch Jahreszahlen zu numeric machen
df_90er["Jahr"] = pd.to_numeric(df_90er["Jahr"])
print(df_90er)
print(df_90er.dtypes)
```

```python
# Daten Wanderung 2000-2025 einlesen
df_2000er = pd.read_csv("daten/wanderungen_2000-2025.csv", sep=";", encoding="utf-8", skiprows=6, header=None)
print(df_2000er.head(10))
```

```python
# Nur die benötigten Spalten in df speichern
df_2000er = df_2000er.iloc[:, [0,13,15,17]]
print(df_2000er.head(10))
```

```python
# Nur die benötigten Zeilen in df speichern
df_2000er = df_2000er.iloc[1:,:]
df_2000er = df_2000er.drop(df_2000er.index[1])
print(df_2000er.head(10))
```

```python
# Erste Datenzeile zu Spaltennamen machen und dann Doppelung entfernen
df_2000er.columns = df_2000er.iloc[0]
df_2000er = df_2000er.iloc[1:].reset_index(drop=True)
df_2000er.columns.name = None
print(df_2000er)
```

```python
# Hinweis und Zeilen mit NaN am Ende entfernen
df_2000er = df_2000er.iloc[:-3]
print(df_2000er.tail(5))
```

```python
# Spalten umbenennen
df_2000er.columns = ["Jahr", "Zuzüge", "Fortzüge", "Saldo"]
print(df_2000er.head(10))
```

```python
# Datatypes der Spalten checken 
print(df_2000er.dtypes)
```

```python
# df 2000er Werte zu numeric machen
df_2000er = df_2000er.apply(pd.to_numeric)
print(df_2000er.dtypes)
```

```python
# df 90er und 2000er abgleichen
print(df_90er)
print(df_2000er.head(10))
```

```python
# csv für spätaussiedler einladen
df_siedler_97 = pd.read_csv("daten/spaetaussiedler_1997-2025.csv", sep=";", header=None, skiprows=6)
print(df_siedler_97.head(20))
```

```python
# Nur die benötigten Spalten in df speichern
df_siedler_97 = df_siedler_97.iloc[:, [0,13]]
print(df_siedler_97.head(10))
```

```python
# Erste Datenzeile zu Spaltennamen machen und dann Doppelung entfernen
df_siedler_97.columns = df_siedler_97.iloc[0]
df_siedler_97 = df_siedler_97.iloc[1:].reset_index(drop=True)
df_siedler_97.columns.name = None
print(df_siedler_97)
```

```python
# Hinweis und Zeilen mit NaN am Ende entfernen
df_siedler_97 = df_siedler_97.iloc[:-4]
print(df_siedler_97.tail(5))
```

```python
# Spalten umbenennen
df_siedler_97.columns = ["Jahr", "Spätaussiedler"]
print(df_siedler_97.head(10))
```

```python
# Datentypen siedler ansehen
print(df_siedler_97.dtypes)
```

```python
# df siedler Werte zu numeric machen
df_siedler_97 = df_siedler_97.apply(pd.to_numeric, errors="coerce")
df_siedler_97["Spätaussiedler"] = df_siedler_97["Spätaussiedler"].astype("Int64")
print(df_siedler_97.dtypes)
```

```python
df_siedler_91 = pd.read_csv("daten/spaetaussiedler_1991-2018.csv")
print(df_siedler_91)
```

```python
# Leerzeilen in Angaben entfernen und in numeric umwandeln
df_siedler_91["Spätaussiedler"] = pd.to_numeric(
    df_siedler_91["Spätaussiedler"].astype(str).str.replace(r"\s+", "", regex=True)
)
print(df_siedler_91)
```

```python
# Datentyp von siedler 91 checken - passt 
print(df_siedler_91.dtypes)
```

```python
# benötigte Werte aus siedler 91 und 97 zusammenführen in neuen df_siedler
df_siedler = pd.concat([
    df_siedler_91.iloc[0:28],
    df_siedler_97[22:29]
], ignore_index=True)

print(df_siedler)
```

```python
# benötigte Werte aus 90er und 2000er zusammenführen in neuen df_wanderung
df_wanderung = pd.concat([
    df_90er.iloc[0:9],
    df_2000er[0:26]
], ignore_index=True)

print(df_wanderung)
```

```python
# Beide DataFrames anhand des Jahres verbinden
df_gesamt = pd.merge(
    df_wanderung, # left
    df_siedler, # right
    on="Jahr",
    how="left",
    validate="one_to_one"
)

print(df_gesamt)
```

```python
# Bereinigte Zuzüge berechnen
df_gesamt["Zuzüge_bereinigt"] = (
    df_gesamt["Zuzüge"] - df_gesamt["Spätaussiedler"]
)
print(df_gesamt)
```

```python
# Bereinigten Saldo berechnen
df_gesamt["Saldo_bereinigt"] = (
    df_gesamt["Saldo"] - df_gesamt["Spätaussiedler"]
)
print(df_gesamt)
```

```python
# Ergebnis als csv speichern
df_gesamt.to_csv(
    "daten/fin_wanderungen_1991-2025.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)
```
